# Отчёт по итогам аналитики деятельности маркетплейса ООО "Гонь" за 2017-2018 гг.

В ходе анализа деятельности компании в разрезе пользователей маркетплейса, сделанных ими заказов и информации о самих заказах, было установлено, что наш продукт имеет крайне низкую возвращаемость пользователей. 

Разбив пользователей на когорты по месяцам и проанализировав их поведение в продукте можно сделать вывод о том, что менее 1% пользователей делают у нас повторные заказы, и данная картина не меняется в течение всего наблюдаемого срока. Всего за 2 года менее 3% пользователей сделали у нас повторный заказ. (График № 1 в презентации)



В связи с этим можно утверждать, что масштабироваться с таким процентом повторных заказов нельзя, так как это приведёт к убыткам. (График № 2 в презентации) 

Сначала наобходимо повысить привлекательность нашего продукта для совершения пользователями повторных заказов.



Чтобы максимизировать прибыль компании, нужно увеличить следующие продуктовые метрики:

1) Среднее количество заказов на одного клиента. Эта метрика в очень плохом состоянии. Низкие показатели по ней означают, что наш маркетплейс работает как "дырявое ведро" - пользователи делают один заказ и уходят к конкурентам.

2) Количество активных пользователей по месяцам. Этот показатель вырос за последние 2 года наблюдений, однако из-за низкой возвращаемости в продукт, польза от такого прироста минимальна. 

3) Конверсия в оплату новых пользователей. Для исследования данной метрики текущих данных было недостаточно, однако всё-равно она является ключевой для определения полезности продукта для пользователей. Поэтому необходимо всегда обращать внимание на неё.

4) Retention (возвращаемость клиентов в продукт). Самый проблемный показатель из всех текущих. Если удастся исправить ситуацию, то можно будет подумать о масштабировании. Однако, при таком уровне retention маркетплейс будет приносить лишь небольшую прибыль или даже убыток.

5) Средняя сумма заказа. Здесь ситуация за 2 года практически не поменялась. Однако, если учесть инфляцию, то данный показатель должен был, хоть и незначительно, но вырасти. Для полноценного исследования этой метрики на данный момент у меня недостаточно данных. Нужно активно наблюдать.

Текущую динамику по этим метрикам за 2017-2018 гг. отражает график 3 в презентации.




Для внесения изменений в продукте было принято на рассмотрение 3 гипотезы:

1) Если исправим баг в системе процессинга заказов, то клиентам не придётся сталкиваться с проблемой отмены заказа, вследствие чего количество доставленных заказов увеличится. Считаем, что мы таким образом избавимся от всех отмен.

2) Если сократим время до отгрузки заказа, то клиенты перестанут получать свой заказ с запаздыванием, вследствие чего количество заказов увеличится за счёт повторных заказов. 

3) Если создадим новый способ оплаты, который будет конвертировать клиентов в повторный заказ, то клиенты не будут испытывать трудности при оформлении заказа, вследствие чего количество заказов увеличится за счёт повторных заказов тех, кто раньше не делал повторный заказ.

Для определения наиболее приоритетной гипотезы для отработки был применён фреймворк ICE, согласно которому каждой гипотезе присуждаются баллы исходя из её полезности (impact), уверенности в том, что она сработает (confidence) и лёгкости в реализации (ease).

В результате наиболее приоритетной для исполнения оказалась гипотеза 1, которая должна избавить маркетплейс от отмен со стороны пользователей из-за бага в системе, так как при относительно высоких показателях confidence и ease она показала самое высокое прогнозируемое положительное влияние на продукт. 


В ходе исполнения этой гипотезы ожидаются изменения по следующим метрикам:


1) выручка (целевая метрика)

2) среднее количество заказов на пользователя (прокси-метрика)

3) конверсию в оплату

4) конверсия в повторный заказ

При этом последние 2 метрики будут являться индикаторами (guardrail-метрики), что всё идёт правильно, поскольку именно по ним мы сможем определить (в случае их уменьшения), нет ли каких-либо сопутствующих проблем при реализации нашей гипотезы.

Общие рекомендации по продукту:

1) Самое срочное и важное, что необходимо сделать - исправить  баг в системе процессинга заказов, благодаря чему клиентам не придётся сталкиваться с проблемой отмены заказа, вследствие чего количество доставленных заказов увеличится и снизится количество отмен.

2) Сконцентрироваться на таких метриках, как среднее количество заказов на одного клиента, количество активных пользователей по месяцам, конверсия в оплату новых пользователей, retention, средняя сумма заказа. Увеличение каждой из этих метрик будет положительно влиять на продукт в целом.

3) Разобраться с задержками доставки заказов. Это негативно влияет на возвращаемость клиентов.
